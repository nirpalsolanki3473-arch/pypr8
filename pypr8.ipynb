```python
import numpy as np


class DataAnalytics:

    def __init__(self, array=None):
        self.array = np.array(array) if array is not None else None

    # ---------------- PRIVATE METHOD ----------------
    def __validate_array(self):
        if self.array is None:
            print("No array has been created.")
            return False
        return True

    # ---------------- STATIC METHOD ----------------
    @staticmethod
    def display_title():
        print("\n====================================")
        print("       Welcome to NumPy Analyzer")
        print("====================================")

    # ---------------- CLASS METHOD ----------------
    @classmethod
    def from_list(cls, values):
        return cls(np.array(values))

    # ---------------- ARRAY CREATION ----------------
    def create_array(self):

        print("\nSelect the type of array to create:")
        print("1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")

        choice = input("Enter your choice: ")

        try:

            if choice == "1":

                values = list(map(int, input(
                    "Enter elements separated by space: "
                ).split()))

                self.array = np.array(values)

            elif choice == "2":

                rows = int(input("Enter number of rows: "))
                columns = int(input("Enter number of columns: "))

                total = rows * columns

                values = list(map(int, input(
                    f"Enter {total} elements separated by space: "
                ).split()))

                if len(values) != total:
                    print("Incorrect number of elements.")
                    return

                self.array = np.array(values).reshape(rows, columns)

            elif choice == "3":

                layers = int(input("Enter number of layers: "))
                rows = int(input("Enter number of rows: "))
                columns = int(input("Enter number of columns: "))

                total = layers * rows * columns

                values = list(map(int, input(
                    f"Enter {total} elements separated by space: "
                ).split()))

                if len(values) != total:
                    print("Incorrect number of elements.")
                    return

                self.array = np.array(values).reshape(
                    layers, rows, columns
                )

            else:
                print("Invalid choice.")
                return

            print("\nArray created successfully:")
            print(self.array)

            self.index_slice_menu()

        except ValueError:
            print("Please enter valid numbers.")

    # ---------------- INDEXING AND SLICING ----------------
    def index_slice_menu(self):

        while True:

            print("\nChoose an operation:")
            print("1. Indexing")
            print("2. Slicing")
            print("3. Go Back")

            choice = input("Enter your choice: ")

            if choice == "1":
                self.indexing()

            elif choice == "2":
                self.slicing()

            elif choice == "3":
                break

            else:
                print("Invalid choice.")

    def indexing(self):

        if not self.__validate_array():
            return

        try:

            if self.array.ndim == 1:

                index = int(input("Enter index: "))
                print("Element:", self.array[index])

            elif self.array.ndim == 2:

                row = int(input("Enter row index: "))
                column = int(input("Enter column index: "))

                print("Element:", self.array[row, column])

            elif self.array.ndim == 3:

                layer = int(input("Enter layer index: "))
                row = int(input("Enter row index: "))
                column = int(input("Enter column index: "))

                print("Element:", self.array[layer, row, column])

        except IndexError:
            print("Index is out of range.")

        except ValueError:
            print("Please enter valid indexes.")

    def slicing(self):

        if not self.__validate_array():
            return

        try:

            if self.array.ndim == 1:

                start = int(input("Enter start index: "))
                end = int(input("Enter end index: "))

                print("Sliced Array:")
                print(self.array[start:end])

            elif self.array.ndim == 2:

                row_range = input(
                    "Enter row range (start:end): "
                )

                column_range = input(
                    "Enter column range (start:end): "
                )

                rs, re = map(int, row_range.split(":"))
                cs, ce = map(int, column_range.split(":"))

                print("Sliced Array:")
                print(self.array[rs:re, cs:ce])

            elif self.array.ndim == 3:

                layer_range = input(
                    "Enter layer range (start:end): "
                )

                row_range = input(
                    "Enter row range (start:end): "
                )

                column_range = input(
                    "Enter column range (start:end): "
                )

                ls, le = map(int, layer_range.split(":"))
                rs, re = map(int, row_range.split(":"))
                cs, ce = map(int, column_range.split(":"))

                print("Sliced Array:")
                print(self.array[ls:le, rs:re, cs:ce])

        except Exception:
            print("Invalid slicing format.")

    # ---------------- MATHEMATICAL OPERATIONS ----------------
    def mathematical_operations(self):

        if not self.__validate_array():
            return

        print("\nChoose a mathematical operation:")
        print("1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Dot Product")
        print("6. Matrix Multiplication")

        choice = input("Enter your choice: ")

        try:

            if choice in ["1", "2", "3", "4"]:

                values = list(map(float, input(
                    f"Enter {self.array.size} elements separated by space: "
                ).split()))

                if len(values) != self.array.size:
                    print("Incorrect number of elements.")
                    return

                second = np.array(values).reshape(self.array.shape)

                print("\nOriginal Array:")
                print(self.array)

                print("\nSecond Array:")
                print(second)

                if choice == "1":
                    result = self.array + second

                elif choice == "2":
                    result = self.array - second

                elif choice == "3":
                    result = self.array * second

                else:

                    if np.any(second == 0):
                        print("Division by zero is not allowed.")
                        return

                    result = self.array / second

                print("\nResult:")
                print(result)

            elif choice == "5":

                values = list(map(float, input(
                    "Enter elements for second array: "
                ).split()))

                second = np.array(values)

                result = np.dot(
                    self.array.flatten(),
                    second.flatten()
                )

                print("Dot Product:", result)

            elif choice == "6":

                if self.array.ndim != 2:
                    print("Matrix multiplication requires a 2D array.")
                    return

                rows = int(input("Enter rows of second matrix: "))
                columns = int(input("Enter columns of second matrix: "))

                total = rows * columns

                values = list(map(float, input(
                    f"Enter {total} elements: "
                ).split()))

                if len(values) != total:
                    print("Incorrect number of elements.")
                    return

                second = np.array(values).reshape(rows, columns)

                if self.array.shape[1] != second.shape[0]:
                    print(
                        "Matrix multiplication not possible."
                    )
                    print(
                        "Columns of first matrix must equal "
                        "rows of second matrix."
                    )
                    return

                result = np.matmul(self.array, second)

                print("\nFirst Matrix:")
                print(self.array)

                print("\nSecond Matrix:")
                print(second)

                print("\nMatrix Multiplication Result:")
                print(result)

            else:
                print("Invalid choice.")

        except ValueError:
            print("Please enter valid numbers.")

    # ---------------- COMBINE / SPLIT ----------------
    def combine_split(self):

        if not self.__validate_array():
            return

        print("\nChoose an option:")
        print("1. Combine Arrays")
        print("2. Split Array")

        choice = input("Enter your choice: ")

        if choice == "1":
            self.combine()

        elif choice == "2":
            self.split()

        else:
            print("Invalid choice.")

    def combine(self):

        try:

            if self.array.ndim != 2:
                print("Combine example currently supports 2D arrays.")
                return

            values = list(map(int, input(
                f"Enter {self.array.size} elements for second array: "
            ).split()))

            if len(values) != self.array.size:
                print("Incorrect number of elements.")
                return

            second = np.array(values).reshape(self.array.shape)

            print("\nOriginal Array:")
            print(self.array)

            print("\nSecond Array:")
            print(second)

            print("\n1. Vertical Stack")
            print("2. Horizontal Stack")
            print("3. Concatenate")

            choice = input("Enter choice: ")

            if choice == "1":

                result = np.vstack((self.array, second))

            elif choice == "2":

                result = np.hstack((self.array, second))

            elif choice == "3":

                result = np.concatenate((self.array, second))

            else:
                print("Invalid choice.")
                return

            print("\nCombined Array:")
            print(result)

        except ValueError:
            print("Invalid input.")

    def split(self):

        try:

            sections = int(
                input("Enter number of sections: ")
            )

            result = np.array_split(
                self.array,
                sections
            )

            print("\nSplit Arrays:")

            for i, part in enumerate(result, start=1):
                print(f"\nPart {i}:")
                print(part)

        except ValueError:
            print("Please enter a valid number.")

    # ---------------- SEARCH / SORT / FILTER ----------------
    def search_sort_filter(self):

        if not self.__validate_array():
            return

        print("\nChoose an operation:")
        print("1. Search a value")
        print("2. Sort the array")
        print("3. Filter values")

        choice = input("Enter your choice: ")

        if choice == "1":
            self.search()

        elif choice == "2":
            self.sort_array()

        elif choice == "3":
            self.filter_array()

        else:
            print("Invalid choice.")

    def search(self):

        try:

            value = float(input("Enter value to search: "))

            positions = np.where(self.array == value)

            if len(positions[0]) == 0:
                print("Value not found.")

            else:
                print("Value found at:")
                print(positions)

        except ValueError:
            print("Please enter a valid value.")

    def sort_array(self):

        print("\n1. Ascending")
        print("2. Descending")

        choice = input("Enter choice: ")

        if choice == "1":

            result = np.sort(self.array)

        elif choice == "2":

            result = np.sort(self.array)

            result = np.flip(result)

        else:
            print("Invalid choice.")
            return

        print("\nOriginal Array:")
        print(self.array)

        print("\nSorted Array:")
        print(result)

    def filter_array(self):

        try:

            print("\nChoose condition:")
            print("1. Greater than")
            print("2. Less than")
            print("3. Greater than or equal")
            print("4. Less than or equal")
            print("5. Equal to")

            choice = input("Enter choice: ")

            value = float(input("Enter value: "))

            if choice == "1":
                result = self.array[self.array > value]

            elif choice == "2":
                result = self.array[self.array < value]

            elif choice == "3":
                result = self.array[self.array >= value]

            elif choice == "4":
                result = self.array[self.array <= value]

            elif choice == "5":
                result = self.array[self.array == value]

            else:
                print("Invalid choice.")
                return

            print("\nFiltered values:")
            print(result)

        except ValueError:
            print("Please enter a valid number.")

    # ---------------- STATISTICS ----------------
    def statistics(self):

        if not self.__validate_array():
            return

        print("\nChoose an aggregate/statistical operation:")
        print("1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")
        print("6. Minimum")
        print("7. Maximum")
        print("8. Percentile")
        print("9. Correlation")

        choice = input("Enter your choice: ")

        if choice == "1":

            print("Sum:", np.sum(self.array))

        elif choice == "2":

            print("Mean:", np.mean(self.array))

        elif choice == "3":

            print("Median:", np.median(self.array))

        elif choice == "4":

            print(
                "Standard Deviation:",
                np.std(self.array)
            )

        elif choice == "5":

            print(
                "Variance:",
                np.var(self.array)
            )

        elif choice == "6":

            print("Minimum:", np.min(self.array))

        elif choice == "7":

            print("Maximum:", np.max(self.array))

        elif choice == "8":

            try:

                percentile = float(
                    input("Enter percentile (0-100): ")
                )

                if 0 <= percentile <= 100:

                    result = np.percentile(
                        self.array,
                        percentile
                    )

                    print(
                        f"{percentile}th Percentile:",
                        result
                    )

                else:
                    print("Percentile must be between 0 and 100.")

            except ValueError:
                print("Invalid percentile.")

        elif choice == "9":

            try:

                values = list(map(float, input(
                    "Enter elements for second array: "
                ).split()))

                second = np.array(values)

                first = self.array.flatten()

                if len(first) != len(second):
                    print(
                        "Both arrays must have the same number "
                        "of elements."
                    )
                    return

                result = np.corrcoef(
                    first,
                    second
                )

                print("\nCorrelation Coefficient Matrix:")
                print(result)

                print(
                    "\nCorrelation coefficient:",
                    result[0, 1]
                )

            except ValueError:
                print("Invalid input.")

        else:
            print("Invalid choice.")


# =====================================================
#                    MAIN PROGRAM
# =====================================================

DataAnalytics.display_title()

obj = DataAnalytics()

while True:

    print("\nChoose an option:")
    print("1. Create a NumPy Array")
    print("2. Perform Mathematical Operations")
    print("3. Combine or Split Arrays")
    print("4. Search, Sort, or Filter Arrays")
    print("5. Compute Aggregates and Statistics")
    print("6. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":

        obj.create_array()

    elif choice == "2":

        obj.mathematical_operations()

    elif choice == "3":

        obj.combine_split()

    elif choice == "4":

        obj.search_sort_filter()

    elif choice == "5":

        obj.statistics()

    elif choice == "6":

        print("\nThank you for using NumPy Analyzer!")
        break

    else:

        print("Invalid choice. Please try again.")
```
